### Accsessing METs API with Python (Meteorologisk Institutt)

The code below uses *requests* to send an API request and receive MET's weather data. The received data will be on a json format. We have used json to print it in a readable manner with indentaion. As you may observe, the json file we receive has an entry called *timeseries*. This contains the hourly weather forecast, beginning at two hours before the current hour. We have saved the weather data of the current hour in a variable called *current*. 

**You will have to fill in the field of *User agent* marked with <...>, since instead of a key MET requires a proper user agent to identify your application.**

```python
import requests
import json

url = "https://api.met.no/weatherapi/locationforecast/2.0/compact"

params = {              
    "lat": 63.4468,     # Trondheim (you can use the coordinates of other cities as well)
    "lon": 10.4219
}

headers = {
    "User-Agent": "<your chosen application name>/1.0 <your NTNU username>@stud.ntnu.no"
}

response = requests.get(url, params=params, headers=headers)
response.raise_for_status() # If there is an error in the response, this will raise an exception

data = response.json()

current = data["properties"]["timeseries"][2]["data"]["instant"]["details"]

print(json.dumps(data, indent=4))

# HINT for task 2: print(current["air_temperature"])
```
